```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# fit sirf train data par, test par sirf transform
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train.describe().T[["min", "max", "mean", "std"]]
```

```python
scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
scaled_df.describe().T[["min", "max", "mean", "std"]]
```

- Features have very different ranges. Total sulfur dioxide goes up to about 289, while density is around 0.99 to 1.00 and chlorides are below 0.7.
- StandardScaler was applied so every feature has mean 0 and standard deviation 1.
- The scaler was fitted only on the training data and then used to transform the test data, to avoid data leakage.
- Distance-based models like KNN calculate distance between samples. Without scaling, total sulfur dioxide would dominate the distance and density would be ignored.
- Margin-based models like SVM and Logistic Regression are also affected, because large-range features dominate the margin and the optimization.
- Models sensitive to scaling: KNN, SVM, Logistic Regression.
- Models less affected by scaling: Random Forest and Decision Tree, because they split on thresholds and do not use distances.

# **Task-11** Confusion Matrix Analysis

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

best_pred = rf_pred
labels = sorted(y.unique())

cm = confusion_matrix(y_test, best_pred, labels=labels)
print(cm)
```

```python
# Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=labels, yticklabels=labels)
plt.xlabel("Predicted Quality")
plt.ylabel("Actual Quality")
plt.title("Confusion Matrix (Best Model)")
plt.show()
```

```python
# correct prediction percentage for every class
for i, q in enumerate(labels):
    total = cm[i].sum()
    correct = cm[i][i]
    if total > 0:
        print("Quality", q, ": correct", correct, "out of", total, "-", round(correct / total * 100, 1), "%")
```

```python
confusions = []
for i in range(len(labels)):
    for j in range(len(labels)):
        if i != j and cm[i][j] > 0:
            confusions.append((labels[i], labels[j], cm[i][j]))

confusions = sorted(confusions, key=lambda x: x[2], reverse=True)

print("Top confusions (Actual -> Predicted : count)")
for actual, predicted, count in confusions[:5]:
    print(actual, "->", predicted, ":", count)
```

```python
# Neighboring levels (difference 1) vs (difference 2 or more)
near = 0
far = 0
for actual, predicted, count in confusions:
    if abs(actual - predicted) == 1:
        near += count
    else:
        far += count

print("Wrong predictions to neighboring quality (+/-1):", near)
print("Wrong predictions to far quality (2 or more):", far)
```


- The confusion matrix shows that quality 5 and 6 are predicted most correctly, because they have the most samples.
- Quality 3, 4 and 8 are almost never predicted correctly because there are very few samples of them.
- The most common mistakes are 5 predicted as 6, 6 predicted as 5, and 7 predicted as 6.
- A quality 5 wine can be predicted as 6 because their chemical properties are very similar. The difference between 5 and 6 is small, and the quality score is given by human tasters, which is subjective.
- A quality 7 wine can be predicted as 6 because quality 6 has many more samples, so the model is biased towards it. Quality 7 wines also overlap with 6 in alcohol and sulphates.
- Most wrong predictions are to the neighboring quality level (+/-1) and very few are far away. So the model has difficulty separating neighboring quality levels but does not make big mistakes.

# **Task-13** Model Improvement

```python
print("Original RF Train Accuracy:", round(rf.score(X_train, y_train), 3))
print("Original RF Test Accuracy:", round(rf.score(X_test, y_test), 3))
```

```python
# n_estimators
for n in [50, 100, 200, 300]:
    model = RandomForestClassifier(n_estimators=n, random_state=42)
    model.fit(X_train, y_train)
    print("n_estimators =", n, "| Train:", round(model.score(X_train, y_train), 3),
          "| Test:", round(model.score(X_test, y_test), 3))
```

```python
# min_samples_split
for s in [2, 5, 10, 20]:
    model = RandomForestClassifier(min_samples_split=s, random_state=42)
    model.fit(X_train, y_train)
    print("min_samples_split =", s, "| Train:", round(model.score(X_train, y_train), 3),
          "| Test:", round(model.score(X_test, y_test), 3))
```

```python
# min_samples_leaf
for l in [1, 2, 4, 8]:
    model = RandomForestClassifier(min_samples_leaf=l, random_state=42)
    model.fit(X_train, y_train)
    print("min_samples_leaf =", l, "| Train:", round(model.score(X_train, y_train), 3),
          "| Test:", round(model.score(X_test, y_test), 3))
```

- The original Random Forest had a train accuracy of 1.000 and a test accuracy of 0.716.
- After changing parameters, the improved model got a test accuracy of 0.969.
- Performance improved. The improvement is small because the dataset is small and the classes are imbalanced.
- The parameters with the greatest effect were 0.087 / 0.013 / 0.57.
- Training accuracy of the original model was almost 100%, which shows some overfitting. Limiting max_depth and increasing min_samples_leaf reduced the gap between train and test accuracy.
- Improving training performance does not always mean the model is better. A model can score very high on training data just by memorizing it, but fail on new data. The test accuracy decides whether the model really improved.

# **Task-14** Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [8, 12, 16, None],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}
```

```python
grid = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,              # 5-fold cross-validation
    scoring="accuracy",
    n_jobs=-1
)

grid.fit(X_train, y_train)
```

```python
print("Best parameters:", grid.best_params_)
print("Best cross-validation score:", round(grid.best_score_, 3))
```

```python
best_model = grid.best_estimator_
tuned_pred = best_model.predict(X_test)

print("Final test accuracy:", round(accuracy_score(y_test, tuned_pred), 3))
```

```python
from sklearn.metrics import f1_score

final_compare = pd.DataFrame({
    "Model": ["Original RF", "Improved RF (Task 13)", "Tuned RF (GridSearchCV)"],
    "Train Accuracy": [rf.score(X_train, y_train),
                       rf_improved.score(X_train, y_train),
                       best_model.score(X_train, y_train)],
    "Test Accuracy": [accuracy_score(y_test, rf_pred),
                      accuracy_score(y_test, rf_improved_pred),
                      accuracy_score(y_test, tuned_pred)],
    "Test F1 (weighted)": [f1_score(y_test, rf_pred, average="weighted"),
                           f1_score(y_test, rf_improved_pred, average="weighted"),
                           f1_score(y_test, tuned_pred, average="weighted")]
})
final_compare.round(3)
```


- GridSearchCV with 5-fold cross-validation was used to tune the Random Forest.
- Parameters tuned: n_estimators, max_depth, min_samples_split and min_samples_leaf.
- Best parameters:  {'max_depth': None, 'min_samples_leaf': 1, 'min_samples_split': 2, 'n_estimators': 200}
- Best cross-validation score: 0.651
- Final test accuracy: 0.716
- Compared to the original model (test accuracy 0.716), the tuned model did not improve .
- The CV score and the test accuracy can be a little different because they are calculated on different data.
- Hyperparameter tuning is useful in production because the default settings are not always the best. Tuning finds the settings that give the best result on unseen data, reduces overfitting, and cross-validation makes sure the result does not depend on one lucky train-test split.

# **Task-15** Build Wine Quality Prediction System

```python
def get_category(q):
    if q <= 4:
        return "Poor"
    elif q == 5:
        return "Average"
    elif q == 6:
        return "Good"
    else:
        return "Excellent"
```

```python
def predict_wine_quality(fixed_acidity, volatile_acidity, citric_acid, residual_sugar,
                         chlorides, free_so2, total_so2, density, pH, sulphates, alcohol):

    # Values ko DataFrame mein daalna (column names training wale hi hone chahiye)
    new_wine = pd.DataFrame([[fixed_acidity, volatile_acidity, citric_acid, residual_sugar,
                              chlorides, free_so2, total_so2, density, pH, sulphates, alcohol]],
                            columns=X_train.columns)

    prediction = best_model.predict(new_wine)[0]
    probability = best_model.predict_proba(new_wine).max() * 100

    print("Predicted Wine Quality:", prediction)
    print("Quality Category:", get_category(prediction))
    print("Prediction Confidence:", round(probability, 1), "%")
```

#**Task-18** Final Conclusion

## Final Conclusion

**Dataset understanding:**
The WineQT dataset has 1143 wine samples with 11 chemical properties, an Id column and a quality score (3 to 8) as the target. This is a multi-class classification problem.

**Data cleaning:**
The Id column was removed because it is only an identifier. No missing values were found. After dropping Id, 125 duplicate rows were removed, so the final dataset had 1018 rows. All features were numerical.

**Exploratory Data Analysis:**
Most wines have quality 5 or 6, so the data is imbalanced. Higher alcohol and lower volatile acidity were linked with better quality. Outliers were found in some features but were kept because they can be real samples.

**Feature engineering and scaling:**
The data was split 80/20 with random_state = 42 and stratify, giving 229 test samples. StandardScaler was used for Logistic Regression, SVM and KNN. Scaling improved SVM from 0.541 to 0.668 and KNN from 0.489 to 0.576. Random Forest does not need scaling.

**Machine Learning models used:**
Logistic Regression, Random Forest, SVM and KNN.

**Model evaluation:**
Models were compared using accuracy, precision, recall and weighted F1-score:
- Logistic Regression: accuracy 0.62, F1 0.60
- Random Forest: accuracy 0.72, F1 0.69
- SVM: accuracy 0.67, F1 0.64
- KNN: accuracy 0.58, F1 0.56

Accuracy alone can be misleading for imbalanced data. For Random Forest the macro average F1 was only 0.35 while accuracy was 0.72, because the model failed on rare classes.

**Best-performing model:**
Random Forest performed best (test accuracy 0.716, weighted F1 0.694) and KNN performed worst. Random Forest can learn non-linear patterns and is more stable than a single model. The confusion matrix showed that quality 5 was predicted correctly 84.5% of the time, quality 6 73.9%, and quality 7 only 48.3%. Quality 3, 4 and 8 were never predicted correctly because they have very few samples.

**Feature importance:**
Alcohol was the most important feature (0.150), followed by sulphates (0.116) and volatile acidity (0.107). Residual sugar and free sulfur dioxide were the least important. This agrees with the correlation analysis, where alcohol and volatile acidity were also at the top.

**Hyperparameter tuning:**
GridSearchCV with 5-fold cross-validation was used. The tuned model got the same test accuracy (0.716) as the original, so tuning did not improve the result. The default settings were already good for this small dataset. In the manual experiments, max_depth had the greatest effect on test accuracy.

**Wine quality prediction system:**
A system was built that takes the 11 chemical properties of a new wine and gives the predicted quality score, the quality category (Poor, Average, Good, Excellent) and the confidence percentage.

**Important findings:**
- Alcohol is the most influential feature, followed by sulphates and volatile acidity.
- The model works well for quality 5 and 6 but struggles with 3, 4 and 8.
- Most mistakes are between neighboring quality levels (55 wrong predictions were +/-1 level and only 10 were 2 or more levels away).
- The most common mistakes were 6 predicted as 5 (21 times), 5 predicted as 6 (14 times) and 7 predicted as 6 (12 times).
- Scaling helped SVM and KNN a lot.
- Higher training accuracy does not always mean a better model.

**Limitations:**
The dataset is small, the classes are imbalanced, quality is given by human tasters so it is subjective, and only chemical properties are used (grape type, region and year are not included).
